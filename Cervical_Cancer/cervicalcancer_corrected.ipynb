```python
# to check for outliers in the dataset, we can use boxplots to visualize the distribution of numerical columns. Outliers are typically represented as points that fall outside the whiskers of the boxplot.
outlier_cols = [ 
    "Age",
    "Number of sexual partners",
    "First sexual intercourse",
    "Num of pregnancies",
    "Smokes (years)",
    "Smokes (packs/year)",
    "Hormonal Contraceptives (years)",
    "IUD (years)",
    "STDs (number)",
    "STDs: Number of diagnosis",
    "STDs: Time since first diagnosis",
    "STDs: Time since last diagnosis",
] # columns to check for outliers
for col in outlier_cols:
    plt.figure(figsize=(8, 6))
    sns.boxplot(data=data, y=col)
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

outlier_col = [
    "Age",
    "Number of sexual partners",
    "First sexual intercourse",
    "Num of pregnancies",
    "Hormonal Contraceptives (years)",
]

mask = pd.Series(True, index=data.index)

for col in outlier_col:
    # Convert numeric-looking strings to numbers; unparseable values become NaN.
    values = pd.to_numeric(data[col], errors="coerce")

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    mask &= values.between(lower, upper) | values.isna()

data = data.loc[mask].copy()

for col in outlier_col:
    plt.figure(figsize=(8, 6))
    sns.boxplot(data=data, y=col)
    plt.title(f'Boxplot of {col} after outlier removal')
    plt.show()
```
